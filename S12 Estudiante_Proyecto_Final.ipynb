## 🔹 Paso 2: Analizar si el negocio es rentable

### 2.1 Cálculo de KPIs principales

**🎯 Objetivo:** Calcular los indicadores clave del negocio para evaluar ingresos, costos y rentabilidad.

Se usan los 3 datasets (`orders`, `catalog`, `marketing_spend`):

**📊 Parte 1: Rentabilidad del negocio**
- ¿Cuál es el ingreso total (revenue)? 
- ¿Cuál es el costo total? 
- ¿Cuánto se ha invertido en marketing? 
- ¿El negocio es rentable? (calcular profit)  

---

**📈 Parte 2: Comportamiento de ventas**
- ¿Cuál es el ticket promedio por orden? 
- ¿Cuál es la cantidad promedio de productos por orden? 
- ¿Cuál es el producto más vendido?
- ¿Cuánto se ha gastado en marketing por canal? 

```python
# PASO 2: ANÁLISIS DE RENTABILIDAD
# ======================

# Crear una copia para el análisis
profitability = orders.copy()

# Unir ORDERS con CATALOG usando el nombre del producto
profitability = profitability.merge(
    catalog[
        [
            'nombre_producto',
            'costo_unitario'
        ]
    ],
    on='nombre_producto',
    how='left'
)

# Revisar el resultado de la unión
print('Dimensiones de la tabla de rentabilidad:')
print(profitability.shape)

print('\nValores ausentes después de unir el catálogo:')
print(
    profitability[
        [
            'cantidad',
            'costo_unitario',
            'monto_total'
        ]
    ].isna().sum()
)
```

```python
# KPIs DE RENTABILIDAD
# ======================

# Calcular costo total por pedido
profitability['costo_total'] = (
    profitability['cantidad'] *
    profitability['costo_unitario']
)

# Calcular ganancia por pedido
profitability['profit'] = (
    profitability['monto_total'] -
    profitability['costo_total']
)

# KPIs principales
revenue_total = profitability['monto_total'].sum()
costo_total = profitability['costo_total'].sum()
profit_total = profitability['profit'].sum()

# Margen de ganancia
margen_profit = (profit_total / revenue_total) * 100

print(f'Revenue total: ${revenue_total:,.2f}')
print(f'Costo total: ${costo_total:,.2f}')
print(f'Profit total: ${profit_total:,.2f}')
print(f'Margen de ganancia: {margen_profit:.2f}%')
```

```python
# RENTABILIDAD POR SEGMENTOS
# ======================

def rentabilidad_segmento(columna):
    
    resultado = (
        profitability
        .groupby(columna, dropna=False)
        .agg(
            revenue=('monto_total', 'sum'),
            costo=('costo_total', 'sum'),
            profit=('profit', 'sum'),
            pedidos=('id_pedido', 'nunique')
        )
        .reset_index()
    )
    
    resultado['margen_profit_pct'] = (
        resultado['profit'] / resultado['revenue'] * 100
    )
    
    return resultado.sort_values('profit', ascending=False)


# Rentabilidad por país
rentabilidad_pais = rentabilidad_segmento('pais')

# Rentabilidad por categoría
rentabilidad_categoria = rentabilidad_segmento('categoria_producto')

# Rentabilidad por fuente de referencia
rentabilidad_fuente = rentabilidad_segmento('fuente_referencia')


print('RENTABILIDAD POR PAÍS')
display(rentabilidad_pais)

print('\nRENTABILIDAD POR CATEGORÍA')
display(rentabilidad_categoria)

print('\nRENTABILIDAD POR FUENTE DE REFERENCIA')
display(rentabilidad_fuente)
```

```python
# NORMALIZAR NOMBRES DE PAÍS
# ======================

# Estandarizar país en los datasets
orders['pais'] = orders['pais'].str.strip().str.title()
marketing['pais'] = marketing['pais'].str.strip().str.title()

# Actualizar también la tabla usada en rentabilidad
profitability['pais'] = profitability['pais'].str.strip().str.title()

# Recalcular rentabilidad por país
rentabilidad_pais = rentabilidad_segmento('pais')

print('RENTABILIDAD POR PAÍS - CORREGIDA')
display(rentabilidad_pais)
```

```python
# PARTE 1: Totales del funnel
# ======================


query_totals = '''
SELECT
    nombre_evento,
    COUNT(DISTINCT id_usuario) AS usuarios
FROM events
GROUP BY nombre_evento
ORDER BY CASE nombre_evento
    WHEN 'first_visit' THEN 1
    WHEN 'select_item' THEN 2
    WHEN 'add_to_cart' THEN 3
    WHEN 'begin_checkout' THEN 4
    WHEN 'add_payment_info' THEN 5
    WHEN 'purchase' THEN 6
END;
'''

totals = pd.read_sql(query_totals, con=engine)
totals
```

```python
# PARTE 2: Conversiones
# ======================

query_conversion = '''
WITH funnel AS (
    SELECT
        nombre_evento,
        COUNT(DISTINCT id_usuario) AS usuarios,
        CASE nombre_evento
            WHEN 'first_visit' THEN 1
            WHEN 'select_item' THEN 2
            WHEN 'add_to_cart' THEN 3
            WHEN 'begin_checkout' THEN 4
            WHEN 'add_payment_info' THEN 5
            WHEN 'purchase' THEN 6
        END AS etapa
    FROM events
    GROUP BY nombre_evento
),

conversiones AS (
    SELECT
        nombre_evento,
        usuarios,
        etapa,
        LAG(usuarios) OVER (ORDER BY etapa) AS usuarios_etapa_anterior
    FROM funnel
)

SELECT
    nombre_evento,
    usuarios,
    ROUND(
        usuarios * 100.0 /
        FIRST_VALUE(usuarios) OVER (ORDER BY etapa),
        2
    ) AS conversion_desde_inicio_pct,
    ROUND(
        usuarios * 100.0 / usuarios_etapa_anterior,
        2
    ) AS conversion_etapa_pct,
    ROUND(
        100 - (usuarios * 100.0 / usuarios_etapa_anterior),
        2
    ) AS drop_off_pct
FROM conversiones
ORDER BY etapa;
'''

conversion = pd.read_sql(query_conversion, con=engine)
conversion
```

## Conclusiones — Paso 3: Funnel de conversión

El funnel muestra una conversión alta a lo largo del recorrido del usuario. De los 7,796 usuarios que registraron first_visit, 6,240 llegaron a purchase, lo que representa una conversión acumulada aproximada de 80.04%.

El mayor punto de abandono se encuentra entre begin_checkout y add_payment_info, donde la conversión entre etapas fue de 86.71% y el drop-off alcanzó 13.29%.

El resto de las etapas presenta pérdidas menores, especialmente entre add_payment_info y purchase, donde la conversión fue de 99.84%.

También se observa que add_to_cart tiene ligeramente más usuarios únicos que select_item. Esto indica que los eventos observados no siguen una secuencia estrictamente decreciente para todos los usuarios, por lo que esta diferencia debe considerarse una limitación al interpretar el funnel.

## 🔹 Paso 4: Evaluar si los usuarios regresan (retención por cohortes)

**🎯 Objetivo:** Analizar la retención de usuarios para entender si regresan después de registrarse.

**Tablas**

- `users` 
- `user_activity` 

---
1. Se identifica la cohorte de cada usuario según el **mes de registro**.


2. Se calcula la retención semanal: cuántos usuarios **se mantienen activos** en cada semana desde su registro.
   - `retenido_w1`: usuarios activos en la semana 1  
   - `retenido_w2`: usuarios activos en la semana 2  
   - `retenido_w3`: usuarios activos en la semana 3  


3. Se calcula el porcentaje de retención para cada semana, dividiendo los usuarios retenidos entre los clientes iniciales de la cohorte:  
   - `semana_1`: porcentaje de usuarios retenidos en la semana 1  
   - `semana_2`: porcentaje de usuarios retenidos en la semana 2  
   - `semana_3`: porcentaje de usuarios retenidos en la semana 3  

Se revisa que la columna de fecha esté en formato correcto (`DATE`).  
Se realiza la conversión usando: `CAST(fecha_registro AS DATE)`

```python
# Explorar tabla users
# =========================
query_users = '''
SELECT *
FROM users;
'''
users = pd.read_sql(query_users, con=engine)
users.head(3)
```

```python
# Explorar tabla user_activity
# =========================
query_activity = '''
SELECT *
FROM user_activity;
'''

user_activity = pd.read_sql(query_activity, con=engine)

user_activity.head(3)
```

```python
# Retención por cohortes
# ======================

query_cohort_retention_final = '''
WITH cohortes AS (
    SELECT
        id_usuario,
        DATE_TRUNC('month', fecha_registro::date)::date AS cohorte
    FROM users
),

tamano_cohorte AS (
    SELECT
        cohorte,
        COUNT(DISTINCT id_usuario) AS usuarios_cohorte
    FROM cohortes
    GROUP BY cohorte
),

usuarios_activos AS (
    SELECT
        c.cohorte,
        ua.dias_despues_registro,
        COUNT(DISTINCT ua.id_usuario) AS usuarios_activos
    FROM cohortes c
    JOIN user_activity ua
        ON c.id_usuario = ua.id_usuario
    WHERE ua.activo = 1
    GROUP BY
        c.cohorte,
        ua.dias_despues_registro
)

SELECT
    a.cohorte,
    a.dias_despues_registro,
    t.usuarios_cohorte,
    a.usuarios_activos,
    ROUND(
        a.usuarios_activos * 100.0 / t.usuarios_cohorte,
        2
    ) AS retencion_pct
FROM usuarios_activos a
JOIN tamano_cohorte t
    ON a.cohorte = t.cohorte
ORDER BY
    a.cohorte,
    a.dias_despues_registro;
'''

cohorte_final = pd.read_sql(
    query_cohort_retention_final,
    con=engine
)

cohorte_final
```

## Conclusiones — Paso 4: Retención por cohortes

El análisis de cohortes muestra que la retención de usuarios se mantiene relativamente estable durante las primeras cuatro semanas posteriores al registro.

En las cohortes analizadas, la retención se encuentra aproximadamente entre 40% y 44% en los días 7, 14, 21 y 28.

No se observa una disminución continua y pronunciada conforme pasan las semanas. En varias cohortes incluso se presentan ligeros incrementos en algunos períodos, lo que indica variaciones en la actividad de los usuarios entre semanas.

Entre los valores más altos se encuentra la cohorte de febrero en el día 21, con una retención de 43.98%. Por otro lado, uno de los valores más bajos corresponde a la misma cohorte en el día 28, con 39.82%.

En términos generales, alrededor de cuatro de cada diez usuarios registrados permanecen activos en los períodos analizados. Esto sugiere que existe una base recurrente de usuarios, aunque también hay una proporción importante que no presenta actividad en cada período de seguimiento.

## 🔹 Paso 5: Validar si los cambios generan impacto (test estadístico)

🎯 **Objetivo:** Evaluar si la modificación en la UI del checkout impacta la **tasa de conversión de compra**.

---

1. **Analizar el dataset** `experiment_checkout_ui.csv` para identificar la métrica principal **conversion**.
   - La métrica **conversion** es 1 si el usuario completó la compra, 0 si no.    
2. **Plantear la hipótesis estadística**     
3. **Aplicar el test estadístico adecuado** 
4. **Interpretar el resultado**  

---
Hipótesis estadística
   - **H₀ (Hipótesis nula):** No existe una diferencia estadísticamente significativa en la tasa de conversión de compra entre el grupo control y el grupo tratamiento.
   - **H₁ (Hipótesis alternativa):** Existe una diferencia estadísticamente significativa en la tasa de conversión de compra entre el grupo control y el grupo tratamiento.
   
**Test estadístico:** Prueba Z de dos proporciones.
**Nivel de significancia alpha:** α = 0.05.

```python
# PASO 5: EXPERIMENTO A/B
# Cargar dataset
# ======================

url_experiment = 'https://practicum-content.s3.amazonaws.com/datasets/experiment_checkout_ui.csv'

experiment = pd.read_csv(url_experiment)

experiment.head()
```

```python
# COMPARACIÓN CONTROL VS TRATAMIENTO
# ======================

resumen_ab = (
    experiment
    .groupby('variante')
    .agg(
        usuarios=('id_usuario', 'nunique'),
        conversiones=('convirtio', 'sum'),
        tasa_conversion=('convirtio', 'mean')
    )
    .reset_index()
)

# Convertir la tasa a porcentaje
resumen_ab['tasa_conversion_pct'] = (
    resumen_ab['tasa_conversion'] * 100
)

display(resumen_ab)
```

```python
# PRUEBA Z DE DOS PROPORCIONES
# ======================

from statsmodels.stats.proportion import proportions_ztest

# Conversiones de cada grupo
conversiones = [
    experiment.loc[
        experiment['variante'] == 'control',
        'convirtio'
    ].sum(),
    
    experiment.loc[
        experiment['variante'] == 'tratamiento',
        'convirtio'
    ].sum()
]

# Número de usuarios de cada grupo
usuarios = [
    experiment.loc[
        experiment['variante'] == 'control',
        'id_usuario'
    ].nunique(),
    
    experiment.loc[
        experiment['variante'] == 'tratamiento',
        'id_usuario'
    ].nunique()
]

# Prueba Z
z_stat, p_value = proportions_ztest(
    count=conversiones,
    nobs=usuarios
)

alpha = 0.05

print(f'Estadístico Z: {z_stat:.4f}')
print(f'P-value: {p_value:.4f}')

if p_value < alpha:
    print('Resultado: se rechaza H0.')
    print('Existe una diferencia estadísticamente significativa entre las variantes.')
else:
    print('Resultado: no se rechaza H0.')
    print('No hay evidencia suficiente de una diferencia estadísticamente significativa.')
```

**Resultado**
Tasa de conversión del grupo control: 15.69%
Tasa de conversión del grupo tratamiento: 16.29%
Estadístico Z: -0.8133
p-value: 0.4161

Como el p-value (0.4161) es mayor que α (0.05), no se rechaza la hipótesis nula.

**Interpretación y recomendación**

Aunque el grupo tratamiento presentó una tasa de conversión ligeramente superior a la del grupo control, la diferencia observada no es estadísticamente significativa.

Por lo tanto, no existe evidencia suficiente para concluir que la modificación en la interfaz del checkout mejore la conversión de compra.

Se recomienda no implementar el tratamiento de forma general basándose únicamente en este experimento. Si el cambio continúa siendo de interés para el negocio, sería conveniente realizar pruebas adicionales antes de tomar una decisión definitiva.
